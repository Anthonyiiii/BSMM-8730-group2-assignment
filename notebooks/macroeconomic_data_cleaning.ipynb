```python
import pandas as pd
import numpy as np
import zipfile
from pathlib import Path
```

```python
project_dir = Path.home() / "Desktop" / "8730 _ Group Assignment"

raw_dir = project_dir / "macroeconomic"

processed_dir = project_dir / "processed"

processed_dir.mkdir(exist_ok=True)

print("Folder exists:", raw_dir.exists())

for file in raw_dir.iterdir():
    print(file.name)
```

```python
policy_rate = pd.read_csv(
    raw_dir / "Policy_Interest_Rate.csv"
)

bond_5y = pd.read_csv(
    raw_dir / "Canada_5Y_Bond_Yield.csv"
)

bond_10y = pd.read_csv(
    raw_dir / "Canada_10Y_Bond_Yield.csv"
)

interest_datasets = {
    "Policy Interest Rate": policy_rate,
    "Canada 5Y Bond Yield": bond_5y,
    "Canada 10Y Bond Yield": bond_10y
}

for name, df in interest_datasets.items():
    print(name, df.shape)
    display(df.head())
```

```python
for name, df in interest_datasets.items():

    print("=" * 60)
    print("Dataset:", name)
    print("=" * 60)

    print("Shape:", df.shape)

    print("\nDate Range:")
    print("Start:", df["Date"].min())
    print("End:", df["Date"].max())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nDuplicate Dates:")
    print(df["Date"].duplicated().sum())

    print("\nData Types:")
    print(df.dtypes)

    print("\nDescriptive Statistics:")
    print(df.describe())

    print("\n")
```

```python
# Standardize interest rate datasets

cleaned_interest = {}

for name, df in interest_datasets.items():

    df = df.copy()

    # Convert date to datetime
    df["Date"] = pd.to_datetime(df["Date"])

    # Standardize column names
    df.columns = df.columns.str.lower()

    # Sort by date
    df = df.sort_values("date").reset_index(drop=True)

    cleaned_interest[name] = df

    print(name, "- Standardization completed")
```

```python
 policy = cleaned_interest["Policy Interest Rate"]
bond5 = cleaned_interest["Canada 5Y Bond Yield"]
bond10 = cleaned_interest["Canada 10Y Bond Yield"]

policy_dates = set(policy["date"])
bond5_dates = set(bond5["date"])
bond10_dates = set(bond10["date"])

print("5Y and 10Y identical dates:",
      bond5_dates == bond10_dates)

print("Policy dates not in 5Y:",
      len(policy_dates - bond5_dates))

print("5Y dates not in Policy:",
      len(bond5_dates - policy_dates))

print("Policy dates not in 10Y:",
      len(policy_dates - bond10_dates))

print("10Y dates not in Policy:",
      len(bond10_dates - policy_dates))

print("\nDates without 5Y bond yield:")

display(sorted(policy_dates - bond5_dates))
```

```python
interest_rates = policy.merge(
    bond5,
    on="date",
    how="outer",
    validate="one_to_one"
)

interest_rates = interest_rates.merge(
    bond10,
    on="date",
    how="outer",
    validate="one_to_one"
)

interest_rates = interest_rates.sort_values(
    "date"
).reset_index(drop=True)

print("Merged Shape:", interest_rates.shape)

print("\nMissing Values:")
print(interest_rates.isnull().sum())

print("\nDuplicate Dates:")
print(interest_rates["date"].duplicated().sum())

display(interest_rates.head(10))
display(interest_rates.tail(10))
```

```python
print("Missing Policy Rate:")

display(
    interest_rates[
        interest_rates["policy_interest_rate"].isna()
    ]
)

print("Missing Bond Yields:")

display(
    interest_rates[
        interest_rates["canada_5y_bond_yield"].isna()
    ]
)
```

```python
# Create a separate dataset for aligned analysis
interest_rates_cleaned = interest_rates.copy()

rate_columns = [
    "policy_interest_rate",
    "canada_5y_bond_yield",
    "canada_10y_bond_yield"
]

for col in rate_columns:

    # Record whether a value was imputed
    interest_rates_cleaned[col + "_imputed"] = (
        interest_rates_cleaned[col].isna()
    )

    # Forward fill using the latest available observation
    interest_rates_cleaned[col] = (
        interest_rates_cleaned[col].ffill()
    )

print("Cleaned Shape:", interest_rates_cleaned.shape)

print("\nRemaining Missing Values:")
print(interest_rates_cleaned.isnull().sum())

display(interest_rates_cleaned.tail(10))
```

```python
print("Total Records:", len(interest_rates_cleaned))

print("Duplicate Dates:",
      interest_rates_cleaned["date"].duplicated().sum())

print("\nNumber of Imputed Values:")

for col in rate_columns:
    print(
        col,
        interest_rates_cleaned[col + "_imputed"].sum()
    )

print("\nRemaining Missing Values:",
      interest_rates_cleaned.isnull().sum().sum())

print("\nDate Range:")
print(interest_rates_cleaned["date"].min())
print(interest_rates_cleaned["date"].max())
```

```python
# Preserve actual observations without imputation
interest_rates.to_csv(
    processed_dir / "interest_rates_observed.csv",
    index=False,
    date_format="%Y-%m-%d"
)

# Export aligned data for subsequent analysis
interest_rates_cleaned.to_csv(
    processed_dir / "interest_rates_cleaned.csv",
    index=False,
    date_format="%Y-%m-%d"
)

print("Export completed")

for file in processed_dir.glob("*.csv"):
    print(file.name)
```

```python
# Load CPI data directly from ZIP

cpi_zip = raw_dir / "Canada_CPI_raw.zip"

with zipfile.ZipFile(cpi_zip) as z:

    with z.open("18100004.csv") as file:

        cpi_raw = pd.read_csv(
            file,
            low_memory=False
        )

print("CPI Raw Shape:", cpi_raw.shape)

display(cpi_raw.head())
```

```python
# Inspect CPI dataset structure

print("Column Names:")
print(cpi_raw.columns.tolist())

print("\nData Types:")
print(cpi_raw.dtypes)

print("\nMissing Values:")
print(cpi_raw.isnull().sum())

print("\nDate Range:")
print(cpi_raw["REF_DATE"].min())
print(cpi_raw["REF_DATE"].max())
```

```python
print("Available Geographies:")
print(cpi_raw["GEO"].unique())

print("\nProduct Categories:")
print(
    cpi_raw["Products and product groups"]
    .unique()[:30]
)

print("\nUnits of Measure:")
print(cpi_raw["UOM"].unique())

print("\nScalar Factors:")
print(cpi_raw["SCALAR_FACTOR"].unique())
```

```python
# Filter Canadian All-items CPI

cpi = cpi_raw[
    (cpi_raw["GEO"] == "Canada") &
    (cpi_raw["Products and product groups"] == "All-items") &
    (cpi_raw["UOM"] == "2002=100") &
    (cpi_raw["VECTOR"] == "v41690973") &
    (cpi_raw["REF_DATE"] >= "2022-01") &
    (cpi_raw["REF_DATE"] <= "2026-09")
].copy()

print("Filtered CPI Shape:", cpi.shape)

display(cpi.head())
display(cpi.tail())
```

```python
# Select relevant columns

cpi = cpi[[
    "REF_DATE",
    "GEO",
    "VALUE",
    "UOM",
    "VECTOR"
]].copy()

# Rename columns

cpi = cpi.rename(columns={
    "REF_DATE": "date",
    "GEO": "geography",
    "VALUE": "cpi_index",
    "UOM": "base_year",
    "VECTOR": "source_vector"
})

# Convert monthly dates

cpi["date"] = pd.to_datetime(
    cpi["date"],
    format="%Y-%m"
)

cpi = cpi.sort_values("date").reset_index(drop=True)

print("Standardized Shape:", cpi.shape)

display(cpi.head())
```

```python
# Validate monthly date continuity before calculation

expected_dates = pd.date_range(
    start="2022-01-01",
    end="2026-09-01",
    freq="MS"
)

assert cpi["date"].is_unique, "Duplicate CPI months"

assert set(cpi["date"]) == set(expected_dates), \
    "Missing or unexpected CPI months"

assert cpi["cpi_index"].notna().all(), \
    "Missing CPI values"

assert (cpi["cpi_index"] > 0).all(), \
    "Invalid CPI values"

# Calculate inflation rates

cpi["inflation_yoy_pct"] = (
    cpi["cpi_index"].pct_change(
        periods=12,
        fill_method=None
    ) * 100
)

cpi["inflation_mom_pct"] = (
    cpi["cpi_index"].pct_change(
        periods=1,
        fill_method=None
    ) * 100
)

# Keep the target analysis period

cpi_cleaned = cpi[
    cpi["date"] >= "2023-01-01"
].copy().reset_index(drop=True)

print("Final CPI Shape:", cpi_cleaned.shape)

display(cpi_cleaned.head(10))
```

```python
print("Total Records:", len(cpi_cleaned))

print("\nMissing Values:")
print(cpi_cleaned.isnull().sum())

print("\nDuplicate Dates:")
print(cpi_cleaned["date"].duplicated().sum())

print("\nDate Range:")
print(cpi_cleaned["date"].min())
print(cpi_cleaned["date"].max())

print("\nDescriptive Statistics:")

display(cpi_cleaned.describe())
```

```python
cpi_output = processed_dir / "cpi_cleaned.csv"

cpi_cleaned.to_csv(
    cpi_output,
    index=False,
    date_format="%Y-%m-%d"
)

print("Export Successful:", cpi_output.exists())
print("Saved to:", cpi_output)
print("Total Records:", len(cpi_cleaned))
```

```python
# Load retail sales data directly from ZIP

retail_zip = raw_dir / "Canada_Retail_Sales_raw.zip"

with zipfile.ZipFile(retail_zip) as z:

    with z.open("20100067.csv") as file:

        retail_raw = pd.read_csv(
            file,
            low_memory=False
        )

print("Retail Sales Raw Shape:", retail_raw.shape)

display(retail_raw.head())
```

```python
# Inspect retail sales dataset

print("Column Names:")
print(retail_raw.columns.tolist())

print("\nData Types:")
print(retail_raw.dtypes)

print("\nMissing Values:")
print(retail_raw.isnull().sum())

print("\nDate Range:")
print(retail_raw["REF_DATE"].min())
print(retail_raw["REF_DATE"].max())

print("\nDuplicate Rows:")
print(retail_raw.duplicated().sum())
```

```python
# Inspect categorical variables

for col in retail_raw.select_dtypes(
    include=["object", "string"]
).columns:

    if col not in [
        "REF_DATE", "DGUID", "VECTOR",
        "COORDINATE"
    ]:

        print("=" * 50)
        print("Column:", col)
        print("Unique Values:", retail_raw[col].nunique())

        print(retail_raw[col].dropna().unique()[:30])
```

```python
# Filter Canadian total retail sales

retail = retail_raw[
    (retail_raw["GEO"] == "Canada") &
    (retail_raw[
        "North American Industry Classification System (NAICS)"
    ] == "Retail trade [44-45]") &
    (retail_raw["Sales, price and volume"] ==
     "Retail sales in current prices") &
    (retail_raw["UOM"] == "Dollars") &
    (retail_raw["SCALAR_FACTOR"] == "millions") &
    (retail_raw["VECTOR"] == "v1446870151") &
    (retail_raw["REF_DATE"] >= "2022-01") &
    (retail_raw["REF_DATE"] <= "2026-09")
].copy()

print("Filtered Retail Sales Shape:", retail.shape)

display(retail.head())
display(retail.tail())

print("\nData Quality Status:")
print(retail["STATUS"].value_counts(dropna=False))
```

```python
# Select relevant columns

retail = retail[[
    "REF_DATE",
    "GEO",
    "VALUE",
    "UOM",
    "SCALAR_FACTOR",
    "VECTOR",
    "STATUS"
]].copy()

# Rename columns

retail = retail.rename(columns={
    "REF_DATE": "date",
    "GEO": "geography",
    "VALUE": "retail_sales_cad_million",
    "UOM": "unit",
    "SCALAR_FACTOR": "scalar_factor",
    "VECTOR": "source_vector",
    "STATUS": "quality_status"
})

# Convert date format

retail["date"] = pd.to_datetime(
    retail["date"],
    format="%Y-%m"
)

retail["currency"] = "CAD"
retail["seasonally_adjusted"] = True

retail = retail.sort_values(
    "date"
).reset_index(drop=True)

print("Standardized Shape:", retail.shape)

display(retail.head())
```

```python
# Validate monthly date continuity

expected_dates = pd.date_range(
    start="2022-01-01",
    end="2026-09-01",
    freq="MS"
)

assert retail["date"].is_unique, \
    "Duplicate retail sales months"

assert set(retail["date"]) == set(expected_dates), \
    "Missing or unexpected months"

assert retail["retail_sales_cad_million"].notna().all(), \
    "Missing retail sales values"

assert (retail["retail_sales_cad_million"] > 0).all(), \
    "Invalid retail sales values"

# Calculate YoY growth

retail["retail_sales_yoy_pct"] = (
    retail["retail_sales_cad_million"]
    .pct_change(periods=12, fill_method=None) * 100
)

# Calculate MoM growth

retail["retail_sales_mom_pct"] = (
    retail["retail_sales_cad_million"]
    .pct_change(periods=1, fill_method=None) * 100
)

# Keep the target analysis period

retail_cleaned = retail[
    retail["date"] >= "2023-01-01"
].copy().reset_index(drop=True)

print("Final Retail Sales Shape:", retail_cleaned.shape)

display(retail_cleaned.head(10))
display(retail_cleaned.tail())
```

```python
# Final retail sales validation

print("Total Records:", len(retail_cleaned))

print("\nMissing Values:")
print(retail_cleaned.isnull().sum())

print("\nDuplicate Dates:")
print(retail_cleaned["date"].duplicated().sum())

print("\nDate Range:")
print(retail_cleaned["date"].min())
print(retail_cleaned["date"].max())

print("\nQuality Status:")
print(retail_cleaned["quality_status"].value_counts(
    dropna=False
))

print("\nDescriptive Statistics:")

display(retail_cleaned[[
    "retail_sales_cad_million",
    "retail_sales_yoy_pct",
    "retail_sales_mom_pct"
]].describe())
```

```python
retail_output = processed_dir / "retail_sales_cleaned.csv"

retail_cleaned.to_csv(
    retail_output,
    index=False,
    date_format="%Y-%m-%d"
)

print("Export Successful:", retail_output.exists())
print("Saved to:", retail_output)
print("Total Records:", len(retail_cleaned))
```
