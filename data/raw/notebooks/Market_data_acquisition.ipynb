```python
import yfinance as yf
import pandas as pd
from pathlib import Path

tickers = {
    "SmartCentres": "SRU-UN.TO",
    "RioCan": "REI-UN.TO",
    "Choice_Properties": "CHP-UN.TO",
    "REIT_Benchmark": "XRE.TO"
}

start_date = "2023-01-01"
end_date = "2026-10-01"

output_dir = Path("market_data")
output_dir.mkdir(exist_ok=True)

for company, ticker in tickers.items():

    data = yf.Ticker(ticker).history(
        start=start_date,
        end=end_date,
        interval="1d",
        auto_adjust=False,
        actions=True
    )

    if data.empty:
        print(f"{company}: Download failed")
        continue

    data.to_csv(output_dir / f"{company}_daily.csv")

    print(f"{company}: {len(data)} rows downloaded")
```

```python
for file in output_dir.glob("*.csv"):

    df = pd.read_csv(file)

    print("File:", file.name)
    print("Rows:", len(df))
    print("Start Date:", df["Date"].min())
    print("End Date:", df["Date"].max())
    print("Missing Values:", df.isnull().sum().sum())
    print("Distribution Records:", (df["Dividends"] > 0).sum())
    print("-" * 50)
```
