```python
import requests
import pandas as pd
from pathlib import Path

series = {
    "Policy_Interest_Rate": "V39079",
    "Canada_5Y_Bond_Yield": "BD.CDN.5YR.DQ.YLD",
    "Canada_10Y_Bond_Yield": "BD.CDN.10YR.DQ.YLD"
}

start_date = "2023-01-01"
end_date = "2026-09-30"

output_dir = Path.home() / "Desktop" / "macroeconomic"
output_dir.mkdir(exist_ok=True)

for name, code in series.items():

    url = f"https://www.bankofcanada.ca/valet/observations/{code}/json"

    params = {
        "start_date": start_date,
        "end_date": end_date
    }

    try:
        response = requests.get(url, params=params, timeout=60)
        response.raise_for_status()

        result = response.json()

        df = pd.DataFrame([
            {
                "Date": row["d"],
                name: row.get(code, {}).get("v")
            }
            for row in result["observations"]
        ])

        if df.empty:
            print(f"{name}: No data returned")
            continue

        df.to_csv(output_dir / f"{name}.csv", index=False)

        print(f"{name}: {len(df)} rows downloaded")

    except requests.RequestException as e:
        print(f"{name}: Download failed - {e}")

print("Saved to:", output_dir)
```

```python

import requests
from pathlib import Path

output_dir = Path.home() / "Desktop" / "macroeconomic"
output_dir.mkdir(exist_ok=True)

datasets = {
    "Canada_CPI_raw": "18100004",
    "Canada_Retail_Sales_raw": "20100067"
}

for name, code in datasets.items():

    api_url = (
        "https://www150.statcan.gc.ca/t1/wds/rest/"
        f"getFullTableDownloadCSV/{code}/en"
    )

    response = requests.get(api_url, timeout=60)
    response.raise_for_status()

    download_url = response.json()["object"]

    result = requests.get(download_url, timeout=120)
    result.raise_for_status()

    file_path = output_dir / f"{name}.zip"
    file_path.write_bytes(result.content)

    print(f"{name}: Download successful")

print("Saved to:", output_dir)
```

```python
import requests
import pandas as pd
from pathlib import Path

output_dir = Path.home() / "Desktop" / "8730 _ Group Assignment" / "macroeconomic"

series = {
    "Policy_Interest_Rate": "V39079",
    "Canada_5Y_Bond_Yield": "BD.CDN.5YR.DQ.YLD",
    "Canada_10Y_Bond_Yield": "BD.CDN.10YR.DQ.YLD"
}

for name, code in series.items():

    url = f"https://www.bankofcanada.ca/valet/observations/{code}/json"

    response = requests.get(url, params={
        "start_date": "2023-01-01",
        "end_date": "2026-09-30"
    }, timeout=60)

    response.raise_for_status()

    df = pd.DataFrame([
        {"Date": row["d"], name: row.get(code, {}).get("v")}
        for row in response.json()["observations"]
    ])

    file_path = output_dir / f"{name}.csv"

    df.to_csv(file_path, index=False)

    print(file_path.name, "Saved:", file_path.exists())
```
